#### 手撕实现一个完整的MLP网络，包含前向传播和反向传播

```python

import numpy as np
from typing import List


class MLP:
    def __init__(self, input_dim: int,
                 hidden_layers: List[int],
                 output_dim: int):
        self.input_dim = input_dim
        self.output_dim = output_dim
        self.hidden_layers = hidden_layers

        self.weight = []
        self.bias = []
        layers = [input_dim] + hidden_layers + [output_dim]

        for i in range(len(layers)-1):
            self.weight.append(np.zeros((layers[i+1],layers[i])))
            self.bias.append(0)
            # 初始化权重矩阵和偏置


    def weight_init(self):
        # 实现权重偏置高斯分布初始化 N(0,1)
        rng = np.random.default_rng()

        for i in range(len(self.weight)):
            self.weight[i] = rng.normal(0, 1, size=self.weight[i].shape)
            self.bias[i] = rng.normal(0, 1)

    def relu(self, x):
        return np.maximum(0, x)

    def forward(self, inputs: np.ndarray) -> np.ndarray:
        # inputs(B, C: input_dim) output:(B, C: output_dim)

        for i in range(len(self.weight)):
            inputs = inputs @ self.weight[i].T + self.bias[i]
            inputs = self.relu(inputs)
        
        return inputs

    def backward(self):
        pass
        
        
```

#### 测试

```python
mlp_1 = MLP(3,[4,4],3)
mlp_1.weight_init()

INPUT = np.array([[0,1,2],
                 [3,4,5]])
# B:2 dim:3

OUTPUT = mlp_1.forward(INPUT)
print(OUTPUT)
```
